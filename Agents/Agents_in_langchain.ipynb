```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
import requests
```

```python
from langchain_community.tools import DuckDuckGoSearchRun
search_tool=DuckDuckGoSearchRun()
results=search_tool.invoke("top news about Ai right now")
```

```python
results
```

```python
from google.colab import userdata
llm=ChatGoogleGenerativeAI(model="gemini-3.8-flash",google_api_key=userdata.get("GOOGLE_API_KEY"))
```

```python
llm.invoke("hi")
```

Main agent working and code

```python
from langchain_classic.agents import create_react_agent,AgentExecutor
from langsmith import Client
client=Client()
prompt=client.pull_prompt( #pulls a standard agent prompt
    "hwchase17/react",
      dangerously_pull_public_prompt=True
)
```

```python
prompt
```

```python
agent=create_react_agent(
    llm=llm,
    tools=[search_tool],
    prompt=prompt
)
```

```python
agent_executor=AgentExecutor(
    agent=agent,
    tools=[search_tool],
    verbose=True
)
```

```python
response=agent_executor.invoke({"input":"3 ways to reach lahore to islamabad"})
print(response)
```
