```python
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from google.colab import userdata
import os

# Securely load your Gemini API Key
os.environ["GOOGLE_API_KEY"] = userdata.get('GOOGLE_API_KEY')

# Initialize Gemini Embeddings
# Using 'models/embedding-001' which is standard for Gemini embeddings
gemini_embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2")
```

```python
from langchain_core.documents import Document

doc1 = Document(
    page_content="Virat Kohli is one of the most successful and consistent batsmen in IPL history. Known for his aggressive batting style and leadership.",
    metadata={"team": "Royal Challengers Bengaluru"}
)

doc2 = Document(
    page_content="Rohit Sharma is one of the most successful captains in IPL history, leading Mumbai Indians to multiple titles. He is known for his calm leadership.",
    metadata={"team": "Mumbai Indians"}
)

doc3 = Document(
    page_content="MS Dhoni is one of the most respected captains in IPL history. He is known for his finishing ability, tactical decisions, and leadership.",
    metadata={"team": "Chennai Super Kings"}
)

doc4 = Document(
    page_content="Jasprit Bumrah is one of the best fast bowlers in the IPL. He is known for his accurate yorkers, variations, and ability to perform under pressure.",
    metadata={"team": "Mumbai Indians"}
)

doc5 = Document(
    page_content="Rashid Khan is a highly effective spin bowler in T20 cricket. He is known for his quick leg-spin, economical bowling, and ability to take wickets.",
    metadata={"team": "Gujarat Titans"}
)
```

```python
docs=[doc1,doc2,doc3,doc4,doc5]
```

```python
# Re-running the initialization now that the API key is set
vectore_store = Chroma(
    embedding_function=gemini_embeddings,
    persist_directory="chroma_db",
    collection_name="sample"
)

# Add the documents to the store
vectore_store.add_documents(docs)
# print("Documents successfully added to the vector store!")
```

```python
vectore_store.get(include=['embeddings', 'documents','metadatas'])
```

```python
vectore_store.similarity_search(
    query='give me only bowler names',
)
```
